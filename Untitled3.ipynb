```python
"""
Pipeline on-chain de swaps POL/USDC (Uniswap V3) con retornos forward via yfinance.
"""

from web3 import Web3
from web3.middleware import ExtraDataToPOAMiddleware
import json
import pandas as pd
import numpy as np
import yfinance as yf
import time
from pathlib import Path
from datetime import datetime, timezone, timedelta

# ──────────────────────────────────────────────────────────────────────────────
# CONSTANTES
# ──────────────────────────────────────────────────────────────────────────────

ZONA_COLOMBIA = timezone(timedelta(hours=-5))
TICKER_POL = "POL-USD"

INTERVALOS_YF = {
    "1m": timedelta(minutes=1),
    "2m": timedelta(minutes=2),
    "5m": timedelta(minutes=5),
    "15m": timedelta(minutes=15),
    "30m": timedelta(minutes=30),
    "1h": timedelta(hours=1),
    "1d": timedelta(days=1),
}

HISTORIA_MAX_DIAS = {
    "1m": 7,
    "2m": 60,
    "5m": 60,
    "15m": 60,
    "30m": 60,
    "1h": 730,
    "1d": 99999,
}

POOL_ABI_SWAP = json.loads('''[{
    "anonymous": false,
    "inputs": [
        {"indexed": true, "internalType": "address", "name": "sender", "type": "address"},
        {"indexed": true, "internalType": "address", "name": "recipient", "type": "address"},
        {"indexed": false, "internalType": "int256", "name": "amount0", "type": "int256"},
        {"indexed": false, "internalType": "int256", "name": "amount1", "type": "int256"},
        {"indexed": false, "internalType": "uint160", "name": "sqrtPriceX96", "type": "uint160"},
        {"indexed": false, "internalType": "uint128", "name": "liquidity", "type": "uint128"},
        {"indexed": false, "internalType": "int24", "name": "tick", "type": "int24"}
    ],
    "name": "Swap",
    "type": "event"
}]''')

# ──────────────────────────────────────────────────────────────────────────────
# 1. CONEXIÓN
# ──────────────────────────────────────────────────────────────────────────────

def conectar_polygon(rpc_url: str) -> Web3:
    w3 = Web3(Web3.HTTPProvider(rpc_url))
    w3.middleware_onion.inject(ExtraDataToPOAMiddleware, layer=0)
    return w3


def obtener_contrato_pool(w3: Web3, pool_address: str):
    return w3.eth.contract(
        address=w3.to_checksum_address(pool_address),
        abi=POOL_ABI_SWAP
    )

# ──────────────────────────────────────────────────────────────────────────────
# 2. DESCARGA DE SWAPS
# ──────────────────────────────────────────────────────────────────────────────

def _bloque_por_timestamp(w3: Web3, ts_objetivo: int, bloque_max: int) -> int:
    lo, hi = 0, bloque_max

    while lo < hi:
        mid = (lo + hi + 1) // 2
        ts_mid = w3.eth.get_block(mid)["timestamp"]

        if ts_mid <= ts_objetivo:
            lo = mid
        else:
            hi = mid - 1

    return lo


def descargar_swaps(
    w3: Web3,
    pool_contract,
    horas: float,
    paso: int = 10,
    pausa_seg: float = 0.05,
    verbose: bool = True,
) -> list:

    ahora_utc = datetime.now(timezone.utc)
    ts_inicio_dt = ahora_utc - timedelta(hours=horas)
    ts_inicio = int(ts_inicio_dt.timestamp())
    bloque_fin = w3.eth.block_number

    if verbose:
        print(
            f"[descarga] buscando bloque para "
            f"{ts_inicio_dt.strftime('%Y-%m-%d %H:%M:%S')} UTC…"
        )

    bloque_ini = _bloque_por_timestamp(w3, ts_inicio, bloque_fin)
    n_bloques = bloque_fin - bloque_ini

    if verbose:
        print(
            f"[descarga] bloques {bloque_ini} → {bloque_fin} "
            f"({n_bloques} bloques / {horas:.2f} h)"
        )

    eventos = []
    total_lotes = max(1, int(np.ceil(n_bloques / paso)))
    lote_n = 0

    for inicio in range(bloque_ini, bloque_fin + 1, paso):
        fin = min(inicio + paso - 1, bloque_fin)

        try:
            batch = pool_contract.events.Swap.get_logs(
                from_block=inicio,
                to_block=fin
            )
            eventos.extend(batch)

        except Exception as e:
            print(f"[error] lote {inicio}-{fin}: {e}")

        lote_n += 1

        if verbose and (lote_n % 50 == 0 or lote_n == total_lotes):
            print(
                f"  lote {lote_n}/{total_lotes} "
                f"({lote_n / total_lotes * 100:.0f}%) — "
                f"swaps: {len(eventos)}"
            )

        time.sleep(pausa_seg)

    if verbose:
        print(f"[descarga] total swaps encontrados: {len(eventos)}")

    return eventos

# ──────────────────────────────────────────────────────────────────────────────
# 3. TABLA TRANSACCIONAL
# ──────────────────────────────────────────────────────────────────────────────

def _enriquecer_evento(
    w3,
    swap,
    decimales_t0=18,
    decimales_t1=6,
    zona=ZONA_COLOMBIA
):
    args = swap["args"]
    n_bloque = swap["blockNumber"]
    hash_tx = swap["transactionHash"].hex()

    if not hash_tx.startswith("0x"):
        hash_tx = "0x" + hash_tx

    try:
        bloque = w3.eth.get_block(n_bloque)
        ts_unix = bloque["timestamp"]
        ts_dt = datetime.fromtimestamp(ts_unix, zona)

        tx = w3.eth.get_transaction(hash_tx)
        recibo = w3.eth.get_transaction_receipt(hash_tx)

        gas_pol = (
            recibo["gasUsed"]
            * recibo.get("effectiveGasPrice", tx.get("gasPrice", 0))
        ) / 1e18

        wallet = tx["from"]

    except Exception as e:
        print(f"[warn] {hash_tx[:14]}… RPC error: {e}")
        ts_unix = 0
        ts_dt = None
        gas_pol = np.nan
        wallet = "error"

    pol_raw = args["amount0"] / 10 ** decimales_t0
    usdc_raw = args["amount1"] / 10 ** decimales_t1

    direccion = (
        "Buy" if pol_raw < 0
        else "Sell" if pol_raw > 0
        else "Unknown"
    )

    # Cantidades siempre positivas.
    pol_cantidad = abs(pol_raw)
    usdc_cantidad = abs(usdc_raw)

    precio_exec = (
        usdc_cantidad / pol_cantidad
        if pol_cantidad > 0
        else np.nan
    )

    return {
        "timestamp": ts_dt,
        "timestamp_unix": ts_unix,
        "bloque": n_bloque,
        "hash_tx": hash_tx,
        "wallet": wallet,
        "direccion": direccion,
        "pol_cantidad": pol_cantidad,
        "usdc_cantidad": usdc_cantidad,
        "precio_ejecutado": precio_exec,
        "gas_pol": gas_pol,
    }


def construir_tabla_swaps(
    w3,
    eventos,
    decimales_t0=18,
    decimales_t1=6,
    zona=ZONA_COLOMBIA,
    verbose=True
):

    filas = []

    for i, swap in enumerate(eventos, 1):

        if verbose and (i % 20 == 0 or i == len(eventos)):
            print(f"  enriqueciendo {i}/{len(eventos)}…")

        filas.append(
            _enriquecer_evento(
                w3,
                swap,
                decimales_t0,
                decimales_t1,
                zona
            )
        )

    df = pd.DataFrame(filas)

    if df.empty:
        return df

    # Gas expresado en USDC.
    df["gas_usdc"] = df["gas_pol"] * df["precio_ejecutado"]

    # Cantidad neta de USDC.
    # Siempre positiva: cantidad de USDC menos gas.
    df["cantidad_neta_usdc"] = (
        df["usdc_cantidad"] - df["gas_usdc"]
    ).clip(lower=0)

    cols = [
        "timestamp",
        "bloque",
        "hash_tx",
        "wallet",
        "direccion",
        "pol_cantidad",
        "usdc_cantidad",
        "precio_ejecutado",
        "gas_pol",
        "gas_usdc",
        "cantidad_neta_usdc"
    ]

    return (
        df[cols]
        .sort_values("timestamp")
        .reset_index(drop=True)
    )

# ──────────────────────────────────────────────────────────────────────────────
# 4. CACHE
# ──────────────────────────────────────────────────────────────────────────────

def guardar_cache(df: pd.DataFrame, cache_path: str) -> None:
    path = Path(cache_path)
    path.parent.mkdir(parents=True, exist_ok=True)

    df.to_parquet(path, index=False)

    print(
        f"[cache] guardado → {path} ({len(df)} swaps)"
    )


def cargar_cache(cache_path: str) -> pd.DataFrame | None:

    path = Path(cache_path)

    if not path.exists():
        return None

    df = pd.read_parquet(path)

    # Compatibilidad con cache anterior.
    if "cantidad_neta_usdc" not in df.columns:

        if "usdc_neto" in df.columns:
            df["cantidad_neta_usdc"] = df["usdc_neto"].abs()

        elif "usdc_cantidad" in df.columns and "gas_usdc" in df.columns:
            df["cantidad_neta_usdc"] = (
                df["usdc_cantidad"] - df["gas_usdc"]
            ).clip(lower=0)

        else:
            raise ValueError(
                "El cache no contiene las columnas necesarias para "
                "calcular cantidad_neta_usdc."
            )

    # Si existe el nombre viejo, se elimina.
    if "usdc_neto" in df.columns:
        df = df.drop(columns=["usdc_neto"])

    print(
        f"[cache] cargado ← {path} ({len(df)} swaps)"
    )

    return df


def _desde_alchemy(
    rpc_url: str,
    pool_address: str,
    horas: float,
    paso: int,
    pausa_seg: float,
    decimales_t0: int,
    decimales_t1: int,
    zona: timezone,
    cache_path: str | None,
    verbose: bool,
) -> pd.DataFrame:

    w3 = conectar_polygon(rpc_url)
    pool = obtener_contrato_pool(w3, pool_address)

    eventos = descargar_swaps(
        w3,
        pool,
        horas=horas,
        paso=paso,
        pausa_seg=pausa_seg,
        verbose=verbose
    )

    df = construir_tabla_swaps(
        w3,
        eventos,
        decimales_t0=decimales_t0,
        decimales_t1=decimales_t1,
        zona=zona,
        verbose=verbose
    )

    if cache_path is not None:
        guardar_cache(df, cache_path)

    return df

# ──────────────────────────────────────────────────────────────────────────────
# 5. PRECIOS FORWARD
# ──────────────────────────────────────────────────────────────────────────────

def _descargar_velas(
    intervalo: str,
    horas_ventana: float
) -> pd.DataFrame:

    TICKER_POL = "POL28321-USD"

    delta_forward = INTERVALOS_YF[intervalo]
    horas_forward = delta_forward.total_seconds() / 3600

    horas_necesarias = (
        horas_ventana + horas_forward
    ) * 1.10

    dias_necesarios = horas_necesarias / 24

    dias_req = min(
        dias_necesarios,
        HISTORIA_MAX_DIAS[intervalo]
    )

    ahora = datetime.now(timezone.utc)
    inicio = ahora - timedelta(
        days=max(dias_req, 1)
    )

    velas = yf.download(
        TICKER_POL,
        start=inicio,
        end=ahora,
        interval=intervalo,
        progress=False,
        auto_adjust=True
    )

    if velas.empty:
        return pd.DataFrame()

    if isinstance(velas.columns, pd.MultiIndex):
        velas.columns = velas.columns.get_level_values(0)

    idx = velas.index

    velas.index = (
        idx.tz_localize("UTC")
        if idx.tz is None
        else idx.tz_convert("UTC")
    )

    return velas[["Close"]].dropna()


def _precio_forward(
    ts_utc: pd.Timestamp,
    delta: timedelta,
    velas: pd.DataFrame
) -> float:

    objetivo = ts_utc + delta

    candidatos = velas[
        velas.index <= objetivo
    ]

    if candidatos.empty:
        return np.nan

    return float(
        candidatos["Close"].iloc[-1]
    )


def agregar_retornos_forward(
    df: pd.DataFrame,
    horizontes: list[str],
    horas_ventana: float,
    verbose: bool = True,
) -> pd.DataFrame:

    invalidos = [
        iv for iv in horizontes
        if iv not in INTERVALOS_YF
    ]

    if invalidos:
        raise ValueError(
            f"Intervalos no soportados: {invalidos}. "
            f"Válidos: {list(INTERVALOS_YF)}"
        )

    df = df.copy()
    df["_ts"] = pd.to_datetime(
        df["timestamp"],
        utc=True
    )

    for iv in horizontes:

        if verbose:
            print(
                f"[yfinance] descargando velas {iv}…"
            )

        velas = _descargar_velas(
            iv,
            horas_ventana
        )

        delta = INTERVALOS_YF[iv]

        if velas.empty:

            if verbose:
                print(f"  [warn] sin datos para {iv}")

            df[f"precio_fwd_{iv}"] = np.nan
            df[f"retorno_{iv}"] = np.nan
            df[f"ganancia_usdc_{iv}"] = np.nan

            continue

        df[f"precio_fwd_{iv}"] = df["_ts"].apply(
            lambda ts: _precio_forward(
                ts,
                delta,
                velas
            )
        )

        # Buy:
        #   gana si POL sube.
        #
        # Sell:
        #   gana si POL baja.
        signo = np.where(
            df["direccion"] == "Buy",
            1,
            -1
        )

        df[f"retorno_{iv}"] = (
            signo
            * (
                df[f"precio_fwd_{iv}"]
                - df["precio_ejecutado"]
            )
            / df["precio_ejecutado"]
        )

        # Ganancia forward expresada en USDC.
        df[f"ganancia_usdc_{iv}"] = (
            df[f"retorno_{iv}"]
            * df["usdc_cantidad"]
        )

    return df.drop(columns=["_ts"])

# ──────────────────────────────────────────────────────────────────────────────
# 6. VISTA POR WALLET
# ──────────────────────────────────────────────────────────────────────────────

def vista_por_wallet(
    df: pd.DataFrame,
    horizontes: list[str]
) -> pd.DataFrame:

    df = df.copy()

    # Compatibilidad adicional por si llega un dataframe antiguo.
    if "cantidad_neta_usdc" not in df.columns:

        if "usdc_neto" in df.columns:
            df["cantidad_neta_usdc"] = (
                df["usdc_neto"].abs()
            )

        else:
            df["cantidad_neta_usdc"] = (
                df["usdc_cantidad"]
                - df["gas_usdc"]
            ).clip(lower=0)

    agg = {
        "n_swaps": (
            "hash_tx",
            "count"
        ),

        "cantidad_pol": (
            "pol_cantidad",
            "sum"
        ),

        "cantidad_usdc": (
            "usdc_cantidad",
            "sum"
        ),

        "gas_usdc": (
            "gas_usdc",
            "sum"
        ),

        "cantidad_neta_usdc": (
            "cantidad_neta_usdc",
            "sum"
        )
    }

    for iv in horizontes:

        col = f"ganancia_usdc_{iv}"

        if col in df.columns:
            agg[f"ganancia_usdc_{iv}"] = (
                col,
                "sum"
            )

    resultado = (
        df.groupby(
            ["wallet", "direccion"]
        )
        .agg(**agg)
        .reset_index()
    )

    # La ganancia forward ya tiene signo económico:
    # positivo = la dirección del swap fue acertada.
    #
    # Se descuenta el gas para obtener la ganancia neta.
    for iv in horizontes:

        bruta_col = f"ganancia_usdc_{iv}"
        neta_col = f"ganancia_neta_usdc_{iv}"

        if bruta_col in resultado.columns:

            resultado[neta_col] = (
                resultado[bruta_col]
                - resultado["gas_usdc"]
            )

            resultado.drop(
                columns=[bruta_col],
                inplace=True
            )

    return (
        resultado
        .sort_values(
            ["wallet", "direccion"]
        )
        .reset_index(drop=True)
    )

# ──────────────────────────────────────────────────────────────────────────────
# 7. PIPELINE COMPLETO
# ──────────────────────────────────────────────────────────────────────────────

def ejecutar_pipeline(
    rpc_url: str,
    pool_address: str,
    horas: float = 10.0,
    horizontes: list[str] = None,
    cache_path: str | None = None,
    forzar_descarga: bool = False,
    paso: int = 10,
    pausa_seg: float = 0.05,
    decimales_t0: int = 18,
    decimales_t1: int = 6,
    zona: timezone = ZONA_COLOMBIA,
    verbose: bool = True,
) -> tuple[pd.DataFrame, pd.DataFrame]:

    if horizontes is None:
        horizontes = [
            "1m",
            "2m",
            "5m",
            "15m",
            "30m",
            "1h"
        ]

    invalidos = [
        iv for iv in horizontes
        if iv not in INTERVALOS_YF
    ]

    if invalidos:
        raise ValueError(
            f"Intervalos no válidos: {invalidos}. "
            f"Opciones: {list(INTERVALOS_YF)}"
        )

    # ── Cache / Alchemy ───────────────────────────────────────────────────────

    tabla_base = None

    if (
        not forzar_descarga
        and cache_path is not None
    ):
        tabla_base = cargar_cache(
            cache_path
        )

    if tabla_base is None:

        tabla_base = _desde_alchemy(
            rpc_url=rpc_url,
            pool_address=pool_address,
            horas=horas,
            paso=paso,
            pausa_seg=pausa_seg,
            decimales_t0=decimales_t0,
            decimales_t1=decimales_t1,
            zona=zona,
            cache_path=cache_path,
            verbose=verbose
        )

    # ── Retornos forward ──────────────────────────────────────────────────────

    tabla_swaps = agregar_retornos_forward(
        tabla_base,
        horizontes=horizontes,
        horas_ventana=horas,
        verbose=verbose
    )

    # ── Vista wallets ─────────────────────────────────────────────────────────

    tabla_wallet = vista_por_wallet(
        tabla_swaps,
        horizontes=horizontes
    )

    return tabla_swaps, tabla_wallet

# ──────────────────────────────────────────────────────────────────────────────
# 8. EJEMPLO DE USO
# ──────────────────────────────────────────────────────────────────────────────

if __name__ == "__main__":

    RPC_URL = "https://polygon-mainnet.g.alchemy.com/v2/alch_C0KE9IzZP8RqJnRzjbxML"

    POOL_ADDRESS = (
        "0xA374094527e1673A86dE625aa59517c5dE346d32" # WPOL /USDC Uniswap,    173k Liquidity
        #0x9B08288C3Be4F62bbf8d1C20Ac9C5e6f9467d8B7  #WPOL /USDT0 Uniswap     1.3M Liquidity
    )

    swaps, wallets = ejecutar_pipeline(
        rpc_url=RPC_URL,
        pool_address=POOL_ADDRESS,
        horas=24,
        cache_path="swaps_24h.parquet",
        verbose=True
    )

    print("\n── TABLA TRANSACCIONAL ──")
    print(
        swaps.head().to_string()
    )

    print("\n── VISTA POR WALLET ──")
    print(
        wallets.head(20).to_string()
    )
```
